```python
def is_prime(n: int) -> bool:
    """
    Arguments: user enter int number
    Returns: True if the number is prime, False otherwise
    """
    if n <= 1:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True


def twin_primes() -> None:
    """
    Arguments: no arguments
    Returns: print twin primes less than 1000
    """
    for n in range(3, 1000, 2):
        if is_prime(n) and is_prime(n + 2):
            print(n, "and", n + 2)

```

```python
twin_primes()
```

```python
def primeFactors(n: int) -> list:
    """
    Arguments: user enter int number
    Returns: list of prime factors of the number
    """
    factors = []

    i = 2

    while n > 1:
        if n % i == 0:
            factors.append(i)
            n = n // i
        else:
            i += 1

    return factors
```

```python
print(primeFactors(56))
```

```python
def dec_to_bin (n: int) -> str:
    '''
    Arguments: user enter int number decimal
    Returns: binary of the number
    '''
    result = ''
    while n > 0:
        result = str(n % 2) + result
        n = n // 2
    return result
```

```python
dec_to_bin(10)
```

```python
def perfectNums(start, end) -> None:
    """
    - argument:
        start: starting number
        end: ending number

    - return:
        None
    """

    for num in range(start, end + 1):
        sum_divisors = 0

        for i in range(1, num):
            if num % i == 0:
                sum_divisors += i

        if sum_divisors == num:
            print(num)





```

```python
perfectNums(0, 100)
```
