```python
class LinearRegressionGD:
    
    def __init__(self, learning_rate=0.001, n_iters=100):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.theta_0 = 0
        self.theta_1 = 0
        self.sse_history = []

    def predict(self, X):
        return self.theta_0 + self.theta_1 * X

    def mse(self, y, y_pred):
        return np.mean((y - y_pred) ** 2)

    def sse(self, y, y_pred):
        return np.sum((y - y_pred) ** 2)

    def fit(self, X, y):
        n = len(X)

        for _ in range(self.n_iters):

           
            y_pred = self.predict(X)

            
            loss = self.sse(y, y_pred)
            self.sse_history.append(loss)

            
            d_theta_0 = (-2 / n) * np.sum(y - y_pred)
            d_theta_1 = (-2 / n) * np.sum(X * (y - y_pred))

            
            self.theta_0 -= self.learning_rate * d_theta_0
            self.theta_1 -= self.learning_rate * d_theta_1

        return self.theta_0, self.theta_1
```

```python
model = LinearRegressionGD(
    learning_rate=0.001,
    n_iters=100
)

model.fit(X, y)
```

```python
print("theta_0:", model.theta_0)
print("theta_1:", model.theta_1)
```

```python
house_size = np.array([70])

prediction = model.predict(house_size)

print("Predicted price:", prediction[0])
```

```python
y_pred = model.predict(X)

print("MSE:", model.mse(y, y_pred))
print("SSE:", model.sse(y, y_pred))
```

```python
plt.plot(model.sse_history)

plt.xlabel("Iterations")
plt.ylabel("SSE")
plt.title("SSE over Iterations")

plt.show()
```

What do X and y represent?

>X represents the house size in square meters (m²).
>y represents the house price in thousands.

Question: 
What do theta_0 and theta_1 represent in the regression equation? 

>y = \theta_0 + \theta_1 X
>θ₀ (theta_0) represents the intercept, which is the predicted value of y when X = 0.
>θ₁ (theta_1) represents the slope, which shows how much y changes when X increases by one unit.

Question: 
Is the prediction reasonable based on the dataset? Why?

>Yes, the prediction is reasonable because 70 m² is within the range of the dataset, and the dataset shows that a 70 m² house has a price of 210 thousand.

 Why SSE decreases over time 

>SSE decreases over time because Gradient Descent updates the model parameters (theta_0 and theta_1) to reduce the prediction error in each iteration.

What convergence means in Gradient Descent

>Convergence in Gradient Descent means that the model parameters become stable and the SSE stops decreasing significantly, indicating that the algorithm has reached or is close to the minimum error.


What happens if the learning rate is too large? 

>If the learning rate is too large, Gradient Descent may overshoot the minimum, causing the SSE to increase or the algorithm to become unstable and fail to converge.
