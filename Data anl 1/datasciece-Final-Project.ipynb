# <a id="1"> 1 ) . What is the distribution for User type?</a>

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5), dpi=200)

sns.countplot(data=df, x='user_type')

plt.xlabel('User Type')
plt.ylabel('Frequency')
plt.title('Distribution of User Type')

plt.show()
```

> ### <mark>Your analsys and exploration here </mark>


The graph shows that Subscribers are much more common than Customers. This means that most users are regular members of the GoBike service. The company could focus on encouraging Customers to become Subscribers.


# <a id="2"> 2 ) What is the distribution for the bike share users on all Trips?</a>

```python
plt.figure(figsize=(10, 5), dpi=200)
sns.countplot(data=df, x='bike_share_for_all_trip')
plt.xlabel('Bike Share for All Trip')
plt.ylabel('Frequency')
plt.title('Distribution of Bike Share for All Trip')
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
Most users did not use bike share for all trips, while only a small number of users used it. This means that the bike share option is not very popular among users, so the company could promote it more.

# <a id="3"> 3 ) What is the distribution for users' ages? </a>

```python
plt.figure(figsize=(12, 6), dpi=200)

sns.histplot(data=df, x='age', bins=30)

plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title("Distribution of Users' Ages")

plt.xticks(range(10, 101, 5))

plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
The age distribution is right-skewed, with most users between 20 and 40 years old. The highest concentration of users is around 30–35 years old. The number of users gradually decreases as age increases, with very few users above 60 years old. Therefore, the company can focus its marketing campaigns mainly on users between 20 and 40 years old.

# <a id="8"> 8) . What the average duration of trip for the categories of user type?</a>

```python
# Distribution of Age by user types
avg_duration = df.groupby('user_type')['duration_houre'].mean()

avg_duration
```

```python
plt.figure(figsize=(8, 5), dpi=200)

sns.barplot(x=avg_duration.index, y=avg_duration.values)

plt.xlabel('User Type')
plt.ylabel('Average Trip Duration (Hours)')
plt.title('Average Trip Duration by User Type')

plt.show()
```

# Make visualization to easy get insight


> ### <mark>Your analsys and exploration here </mark>
The bar chart compares the average trip duration across different user types. It shows that Customers tend to have a longer average trip duration than Subscribers. This suggests that Customers may use the bike-sharing service for longer or more occasional trips, while Subscribers appear to use the service more regularly for shorter trips, possibly for commuting or daily transportation. Overall, user type seems to have an impact on trip duration, with a clear difference in the average duration between the two groups.
