```python
email = input("Enter email: ")

if email.count("@") != 1 or "." not in email.split("@")[1]:
    print("Invalid email")
else:
    username = email.split("@")[0]
    domain = email.split("@")[1].rsplit(".", 1)[0]

    print("Username:", username)
    print("Domain:", domain)

    if email.endswith(".com"):
        print("Commercial Domain")
    elif email.endswith(".edu"):
        print("Educational Domain")
    else:
        print("Other Domain")
```

```python
message = "###!!@mocleW EPGTQ!!!6789"

core = ""

for ch in message:
    if ch.isalpha() or ch == " ":
        core += ch

words = core.split()

first_word = words[0][::-1]
second_word = words[1]

print(first_word, second_word)
```

```python
message = "&&&**$gnirtS PLIO!!@1234"

core = ""

for ch in message:
    if ch.isalpha() or ch == " ":
        core += ch

words = core.split()

first_word = words[0][::-1]

second_word = words[1]
second_word = second_word.replace("I", "E")
second_word = second_word.replace("O", "U")

print(first_word, second_word)
```
