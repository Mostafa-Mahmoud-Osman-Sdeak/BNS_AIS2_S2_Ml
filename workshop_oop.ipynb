```python
import random
import string
```

```python
class NumberTools:
    def factorial(self, num):
        if num == 0 or num == 1:
            return 1
        return num * self.factorial(num - 1)

    def is_prime(self, num):
        if num < 2:
            return False
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return False
        return True

    def common_divisor(self, num1, num2):
        divisors = []
        for i in range(1, min(num1, num2) + 1):
            if num1 % i == 0 and num2 % i == 0:
                divisors.append(i)
        return divisors
```

```python
numbers = NumberTools()

print(numbers.factorial(5))
print(numbers.is_prime(17))
print(numbers.common_divisor(12, 18))
```

```python
class TextTools:
    def reverse_text(self, text):
        return " ".join(text.split()[::-1])

    def reverse_word(self, word):
        return word[::-1]

    def count_word(self, text, word):
        words = text.lower().split()
        return words.count(word.lower())
```

```python
text_tools = TextTools()

print(text_tools.reverse_text("I love Python programming"))
print(text_tools.reverse_word("Python"))
print(text_tools.count_word("Python is easy and Python is powerful", "Python"))
```

```python
class PasswordGenerator:
    def __init__(self, length=8):
        self.length = length

    def generate_pass(self):
        characters = string.ascii_letters + string.digits + string.punctuation
        return ''.join(random.choice(characters) for _ in range(self.length))
```

```python
password = PasswordGenerator(10)
print(password.generate_pass())
```

```python
class SimpleChatbot:
    def __init__(self):
        self.responses = {
            "hello": "Hello! How can I help you?",
            "hi": "Hi! Nice to meet you.",
            "how are you": "I'm fine, thank you!",
            "bye": "Goodbye!,good luck"
        }

    def get_response(self, message):
        message = message.lower().strip()
        return self.responses.get(message, "Sorry, I don't understand.")

    def run(self):
        while True:
            user_message = input("You: ")
            print("Bot:", self.get_response(user_message))
            if user_message.lower().strip() == "bye":
                break
```

```python
bot = SimpleChatbot()
bot.run()
```
