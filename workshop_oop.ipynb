```python
import random
import string
```

```python
class NumberTools:
    def factorial(self, num):
        if num == 0 or num == 1:
            return 1
        return num * self.factorial(num - 1)

    def is_prime(self, num):
        if num < 2:
            return False
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return False
        return True

    def common_divisor(self, num1, num2):
        divisors = []
        for i in range(1, min(num1, num2) + 1):
            if num1 % i == 0 and num2 % i == 0:
                divisors.append(i)
        return divisors
```

```python
numbers = NumberTools()

print(numbers.factorial(5))
print(numbers.is_prime(17))
print(numbers.common_divisor(12, 18))
```

```python
class TextTools:
    def reverse_text(self, text):
        return " ".join(text.split()[::-1])

    def reverse_word(self, word):
        return word[::-1]

    def count_word(self, text, word):
        words = text.lower().split()
        return words.count(word.lower())
```

```python
text_tools = TextTools()

print(text_tools.reverse_text("I love Python programming"))
print(text_tools.reverse_word("Python"))
print(text_tools.count_word("Python is easy and Python is powerful", "Python"))
```

```python
class PasswordGenerator:
    def __init__(self, length=8):
        self.length = length

    def generate_pass(self):
        characters = string.ascii_letters + string.digits + string.punctuation
        return ''.join(random.choice(characters) for _ in range(self.length))
```

```python
password = PasswordGenerator(10)
print(password.generate_pass())
```

```python
class SimpleChatbot:
    def __init__(self):
        self.responses = {
            "hello": "Hello! How can I help you?",
            "hi": "Hi! Nice to meet you.",
            "how are you": "I'm fine, thank you!",
            "bye": "Goodbye!,good luck"
        }

    def get_response(self, message):
        message = message.lower().strip()
        return self.responses.get(message, "Sorry, I don't understand.")

    def run(self):
        while True:
            user_message = input("You: ")
            print("Bot:", self.get_response(user_message))
            if user_message.lower().strip() == "bye":
                break
```

```python
bot = SimpleChatbot()
bot.run()
```

```python
class Calculator:
    """
    Calculator class performs basic arithmetic operations.
    """

    def __init__(self, num1, num2):
        """
        Initialize the calculator with two numbers.
        """
        self.num1 = num1
        self.num2 = num2

    def add(self):
        """
        Returns the addition of two numbers.
        """
        return self.num1 + self.num2

    def subtract(self):
        """
        Returns the subtraction of two numbers.
        """
        return self.num1 - self.num2

    def multiply(self):
        """
        Returns the multiplication of two numbers.
        """
        return self.num1 * self.num2

    def divide(self):
        """
        Returns the division of two numbers.
        """
        if self.num2 == 0:
            return "Cannot divide by zero!"
        return self.num1 / self.num2


print("Welcome to the Simple Calculator!")

while True:

    print("\n1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice not in ["1", "2", "3", "4", "5"]:
        print("Invalid choice. Please try again.")
        continue

    if choice == "5":
        print("Thank you for using the calculator!")
        break

    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    # Create an object
    calculator = Calculator(num1, num2)

    if choice == "1":
        print("Result:", calculator.add())

    elif choice == "2":
        print("Result:", calculator.subtract())

    elif choice == "3":
        print("Result:", calculator.multiply())

    elif choice == "4":
        print("Result:", calculator.divide())

    again = input("Do you want another calculation? (yes/no): ")

    if again.lower() == "no":
        print("Thank you for using the calculator!")
        break
```
